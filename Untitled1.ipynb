```python
import pandas as pd
import numpy as np
import psutil
import time
import joblib

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
import time

data = []

print("Collecting system data...")

for i in range(100):
    cpu = psutil.cpu_percent(interval=1)
    ram = psutil.virtual_memory().percent
    disk = psutil.disk_usage("C:\\").percent

    network = psutil.net_io_counters()

    network_usage_mb = (
        network.bytes_sent + network.bytes_recv
    ) / (1024 * 1024)

    active_processes = len(psutil.pids())

    data.append([
        cpu,
        ram,
        disk,
        network_usage_mb,
        active_processes
    ])

    print(f"Reading {i + 1}/100")

    time.sleep(2)

print("Data collection completed!")
```

```python
df_realtime = pd.DataFrame(
    data,
    columns=[
        "cpu_usage_percent",
        "ram_usage_percent",
        "disk_usage_percent",
        "network_usage_mb",
        "active_processes"
    ]
)

df_realtime.head()
```

```python
print("Shape:", df_realtime.shape)

print("\nMissing Values:")
print(df_realtime.isnull().sum())

print("\nStatistics:")
df_realtime.describe()
```

```python
def calculate_risk(row):

    score = 0

    # CPU
    if row["cpu_usage_percent"] >= 90:
        score += 3
    elif row["cpu_usage_percent"] >= 70:
        score += 2
    elif row["cpu_usage_percent"] >= 50:
        score += 1

    # RAM
    if row["ram_usage_percent"] >= 90:
        score += 3
    elif row["ram_usage_percent"] >= 70:
        score += 2
    elif row["ram_usage_percent"] >= 50:
        score += 1

    # Disk
    if row["disk_usage_percent"] >= 90:
        score += 2
    elif row["disk_usage_percent"] >= 80:
        score += 1

    # Active processes
    if row["active_processes"] >= 200:
        score += 2
    elif row["active_processes"] >= 100:
        score += 1

    if score >= 6:
        return "High"

    elif score >= 3:
        return "Medium"

    else:
        return "Low"
```

```python
df_realtime["failure_risk"] = df_realtime.apply(
    calculate_risk,
    axis=1
)

df_realtime.head(10)
```

```python
df_realtime["failure_risk"].value_counts()
```

```python
X = df_realtime[
    [
        "cpu_usage_percent",
        "ram_usage_percent",
        "disk_usage_percent",
        "network_usage_mb",
        "active_processes"
    ]
]

y = df_realtime["failure_risk"]

print("Features:")
display(X.head())

print("\nTarget:")
display(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

rf_model.fit(
    X_train,
    y_train
)

print("Model training completed!")
```

```python
y_pred = rf_model.predict(X_test)

accuracy = accuracy_score(
    y_test,
    y_pred
)

print(
    f"Model Accuracy: {accuracy * 100:.2f}%"
)
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
joblib.dump(
    rf_model,
    "realtime_failure_prediction_model.pkl"
)

print("✅ Real-time model saved successfully!")
```

```python
df_realtime.to_csv(
    "realtime_system_data.csv",
    index=False
)

print("✅ Dataset saved successfully!")
```
